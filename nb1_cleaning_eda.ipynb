```python
# File is downloaded to the data folder in project directory
file_path = "data/MUP_INP_RY26_P03_V10_DY24_PrvSvc.csv"
raw_df = pd.read_csv(file_path, low_memory=False)
raw_df.head(5)
```

```python
# Data cleaning and preprocessing for pipeline

COLUMN_MAPPING = {
        'Rndrng_Prvdr_CCN': 'provider_ccn',
        'Rndrng_Prvdr_Org_Name': 'provider_name',
        'Rndrng_Prvdr_City': 'city',
        #'Rndrng_Prvdr_St': 'provider_street_address',
        #'Rndrng_Prvdr_State_FIPS': 'provider_state_FIPS',
        #'Rndrng_Prvdr_Zip5': 'provider_zipcode',
        'Rndrng_Prvdr_State_Abrvtn': 'state',
        'Rndrng_Prvdr_RUCA': 'ruca', # RUCA for the zipcode where provider is physically
        'Rndrng_Prvdr_RUCA_Desc': 'ruca_desc',
        'DRG_Cd': 'drg_code',
        'DRG_Desc': 'drg_desc',
        'Tot_Dschrgs': 'total_discharges',
        'Avg_Submtd_Cvrd_Chrg': 'avg_charge',
        'Avg_Tot_Pymt_Amt': 'avg_payment',
        'Avg_Mdcr_Pymt_Amt': 'avg_medicare_payment'
}
def select_and_rename(df):
    """
    Subsets relevant columns and renames to more readable/descriptive names based on mapping.
    """
    # Unused columns commented out, can be added back if needed

    kept_cols = [col for col in COLUMN_MAPPING.keys() if col in df.columns]
    
    return df[kept_cols].rename(columns=COLUMN_MAPPING)


def fix_dtypes(df):
    """
    Fix data types for specific columns 
    (provider_ccn and drg_code handled in subsequent function to pad identifiers)
    """
    df['state'] = df['state'].astype('category')    
    
    return df

def pad_identifiers(df):
    """
    Pad identifiers with leading zeros to ensure consistent length: 
    provider_ccn (6 digits), drg_code (3 digits)
    """
    df['provider_ccn'] = df['provider_ccn'].astype(str).str.zfill(6)
    df['drg_code'] = df['drg_code'].astype(str).str.zfill(3)
    return df

def add_urbanicity_column(df): 
    """
    Add a new column 'urbanicity' based on the provider_ruca code.

    per Wikipedia, 1-3 is Metropolitan, 4-6 is Micropolitan, 7-10 is Small town/Rural (up to 10.6).
    """
    df["urbanicity"] = np.select(
        [df["ruca"].between(1, 3), df["ruca"].between(4, 10.6)],
        ["Metro", "Non-metro"],
        default="Unknown/Invalid",
    )
    df["urbanicity"] = df["urbanicity"].astype("category")
    return df

def add_pricing_metrics(df): 
    """
    realization = avg_medicare_payment / avg_charge 
    markup = avg_charge / avg_medicare_payment
    """
    df["realization"] = df["avg_medicare_payment"] / df["avg_charge"]
    df["markup"] = df["avg_charge"] / df["avg_medicare_payment"]
    return df
```

```python
cleaned_df = (
    raw_df
    .pipe(select_and_rename)
    .pipe(fix_dtypes)
    .pipe(pad_identifiers)
    .pipe(add_urbanicity_column)
    .pipe(add_pricing_metrics)
)

print("Cleaned DataFrame Summary:")
print(f"Rows: {cleaned_df.shape[0]}, Columns: {cleaned_df.shape[1]}")
print(f"Unique states (DC counts as a state): {cleaned_df['state'].nunique()}")
print(f"Unique providers: {cleaned_df['provider_ccn'].nunique()}")
print(f"Unique DRG codes: {cleaned_df['drg_code'].nunique()}")
print(f"Duplicate rows: {cleaned_df.duplicated().sum()}")
print(f"\nUnique urbanicity categories: {cleaned_df['urbanicity'].nunique()}")
print(cleaned_df['urbanicity'].value_counts())
```

```python
# RUCA Code Counts

ruca_df = (
    cleaned_df.groupby(["ruca", "urbanicity"], observed=True, dropna=False)
    .size()
    .reset_index(name="count")
    .sort_values("ruca", ascending=True)
)
ruca_df["ruca_label"] = ruca_df["ruca"].astype(str)

fig = px.bar(
    ruca_df,
    x="count",
    y="ruca_label",
    color="urbanicity",
    orientation="h",
    log_x=True,
    category_orders={"ruca_label": ruca_df["ruca_label"].unique().tolist()},
    labels={"count": "Count (log scale)", "ruca_label": "RUCA Code", "urbanicity": "Urbanicity"},
    title="Distribution of RUCA Codes in Cleaned Data",
)
fig.show()
```

```python
# Distribution of Realization

median_realization = cleaned_df["realization"].median()
average_realization = cleaned_df["realization"].mean()

fig = px.histogram(
    cleaned_df, 
    x=cleaned_df['realization'], 
    range_x=[0, 1.5],
    nbins=160, 
    title='<b>Distribution of Realization Rate</b><br>Share of billed charges actually collected', 
    labels={'realization': 'Realization (Payment / Charge)'})

fig.update_layout(
    xaxis_title='Realization (Payment / Charge)',
    yaxis_title='Hospital-procedure records', 
    bargap=0.1
)
fig.add_vline(
    x=median_realization, 
    line_dash="dash", 
    line_color="red",
    annotation_text=f"Median: {median_realization:.1%}"
)
# fig.add_vline(
#     x=average_realization, 
#     line_dash="dash", 
#     line_color="blue",
#     annotation_text=f"Average: {average_realization:.1%}"
# )

fig.show()
```

```python
# Most common DRG codes by number of discharges 
top_ten_drg = (
    cleaned_df.groupby(['drg_code', 'drg_desc'])
    .agg(
        hospital_count=('provider_ccn', 'nunique'),
        total_discharges=('total_discharges', 'sum'),
        median_charge=('avg_charge', 'median'),
        median_payment=('avg_medicare_payment', 'median'),
        median_realization=('realization', 'median')
    )
    .sort_values("total_discharges", ascending=False)
    .head(10)
)
top_ten_drg
```

```python
# Nunber of unique DRG per hospital/provider
drg_counts = cleaned_df.groupby("provider_ccn")["drg_code"].nunique()
median_drg_count = drg_counts.median()
threshold_drg_count = 10  # cutoff threshold for analysis

fig = px.histogram(
    drg_counts.reset_index(name='unique_drg_count'),
    x='unique_drg_count',
    nbins=50,
    title='<b>Distribution of Unique DRG Codes per Hospital/Provider</b>',
    labels={'unique_drg_count': 'Number of Unique DRG Codes'}
)
fig.update_layout(
    xaxis_title='Number of Unique DRG Codes',
    yaxis_title='Number of Hospitals/Providers', 
    bargap=0.1,
    xaxis=dict(range=[0, 400])
)
fig.add_vline(
    x=median_drg_count,
    line_dash="dash",
    line_color="red",
    annotation_text=f"Median: {median_drg_count}",
    annotation_position="top",
    annotation=dict(yshift=0),
)
fig.add_vline(
    x=threshold_drg_count,
    line_dash="dash",
    line_color="orange",
    annotation_text=f"Threshold: {threshold_drg_count} DRG Codes",
    annotation_position="top",
    annotation=dict(yshift=-22),
)

fig.show()
```
